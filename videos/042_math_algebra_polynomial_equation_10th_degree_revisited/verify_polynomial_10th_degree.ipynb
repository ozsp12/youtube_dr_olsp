# Symbolic verification of a tenth-degree polynomial equation

Computational companion to *A Tenth-Degree Polynomial Equation: Factorization, Roots of Unity, and Cyclotomic Polynomials* by Osvaldo L. Santos-Pereira.

Every computational check uses exact SymPy arithmetic. Finite parameter grids validate instances but do not replace proofs for arbitrary parameters.


## Setup and notation

We import the verification module and define the conventions used throughout the teaching note. The polynomial $P(x)$ is the tenth-degree expression analyzed in the paper and both presentations.

$$
P(x)=1+x+x^2+x^4+x^5+x^6+x^8+x^9+x^{10}.
$$

```python
import sympy as sp
from IPython.display import display, Math
import verify_polynomial_10th_degree as v
from verify_polynomial_10th_degree import x,t,P,S,Q,Pmb,V
print('SymPy version:',sp.__version__)
display(Math(r'P(x)='+sp.latex(P)))
```

## 1. Exponent blocks

The exponents of the nonzero terms are uniquely represented as $r+4s$ with $r,s\in\{0,1,2\}$. This produces a product of two geometric sums without guesswork.

$$
P(x)=\sum_{s=0}^{2}\sum_{r=0}^{2}x^{r+4s}=(1+x+x^2)(1+x^4+x^8).
$$

```python
print(v.check_original_blocks())
```

## 2. Finite geometric sums

The telescoping identity is a polynomial identity valid at every complex value. At $z=1$, evaluate the original sum rather than a quotient with zero denominator. We test $2\le m\le12$ and $z=x^b$ for $1\le b\le8$.

$$
S_m(z)=\sum_{j=0}^{m-1}z^j,\qquad(z-1)S_m(z)=z^m-1,\qquad S_m(1)=m.
$$

```python
print(v.check_finite_geometric_sums())
```

## 3. Orders of roots of unity

The order is the least positive exponent that gives one. For each integer $1\le n\le12$ we verify its indexed order classification and the decomposition of $x^n-1$.

$$
\omega_{n,h}=e^{2\pi ih/n},\quad \operatorname{ord}(\omega_{n,h})=\frac{n}{\gcd(n,h)}.
$$

$$
x^n-1=\prod_{d\mid n}\Phi_d(x),\qquad n=\sum_{d\mid n}\varphi(d).
$$

```python
print(v.check_roots_of_unity())
```

## 4. Cyclotomic polynomials through order 12

Each primitive $n$-th root appears once in the monic cyclotomic polynomial. Its degree is Euler's totient function. The verification also tests irreducibility over the rational numbers.

$$
\Phi_n(x)=\prod_{\substack{0\le h<n\\\gcd(h,n)=1}}(x-\omega_{n,h}),\qquad \deg\Phi_n=\varphi(n).
$$

```python
print(v.check_cyclotomic_polynomials())
```

## 5. Original cyclotomic factorization

The first factor is $\Phi_3$, and the second contributes $\Phi_3\Phi_6\Phi_{12}$. The polynomial gcd with its derivative detects the repeated primitive cubic roots.

$$
P(x)=\Phi_3(x)^2\Phi_6(x)\Phi_{12}(x),\qquad\gcd(P,P')=\Phi_3(x).
$$

```python
print(v.check_original_cyclotomic_factorization())
```

## 6. Multiplicity balance at the twelve candidate roots

The product identity implies a pointwise multiplicity balance. Among the twelve roots of unity, the four axial points are excluded, the two nontrivial cubic roots are double, and six are simple.

$$
(1-x)(1-x^4)P(x)=(1-x^3)(1-x^{12}),
$$

$$
\operatorname{mult}_{P}(\omega_{12,h})=1+\mathbf1_{4\mid h}-\mathbf1_{3\mid h}-\mathbf1_{h=0}.
$$

```python
print(v.check_multiplicity_balance())
```

## 7. Independent substitution

Set $u=x^4$ in the second geometric sum. Solving the auxiliary quadratic produces two arguments, each yielding four fourth roots. The two primitive cube roots also solve the first factor.

$$
1+u+u^2=0,\quad u=e^{\pm 2\pi i/3},
$$

$$
\theta=\pm\frac{\pi}{6}+\frac{k\pi}{2},\qquad k=0,1,2,3.
$$

```python
print(v.check_fourth_power_substitution())
```

## 8. Real quadratic factors

Combining conjugate roots yields real quadratic factors. The two irrational-coefficient factors multiply to $\Phi_{12}$, and all quadratic discriminants are negative.

$$
P(x)=(x^2+x+1)^2(x^2-x+1)(x^2-\sqrt3x+1)(x^2+\sqrt3x+1).
$$

```python
print(v.check_real_factorization())
```

## 9. Viète and symmetric sums

Roots are listed with multiplicities. The module uses exact roots of unity to verify the sum of roots, the second elementary symmetric sum and the product of all ten roots.

$$
e_1=\sum r_j=-1,\qquad e_2=\frac{(\sum r_j)^2-\sum r_j^2}{2}=1,\qquad e_{10}=\prod r_j=1.
$$

```python
print(v.check_viete())
```

## 10. Chebyshev recurrence and symmetric powers

Chebyshev polynomials provide a polynomial representation of the Laurent terms generated by reciprocal reduction. We verify both standard recurrence families and the auxiliary sequence $V_j$.

$$
T_{j+1}(y)=2yT_j(y)-T_{j-1}(y),\qquad U_{j+1}(y)=2yU_j(y)-U_{j-1}(y),
$$

$$
V_j(t)=2T_j(t/2),\qquad V_j(x+x^{-1})=x^j+x^{-j}.
$$

```python
print(v.check_chebyshev_recurrences())
```

## 11. Reciprocal reduction of the tenth-degree polynomial

Since the original polynomial is reciprocal, dividing by $x^5$ and introducing $t=x+x^{-1}$ reduces the problem to a quintic with an elementary factorization.

$$
x^{-5}P(x)=t^5+t^4-4t^3-4t^2+3t+3=(t+1)^2(t-1)(t^2-3).
$$

```python
print(v.check_reciprocal_reduction())
```

## 12. General reciprocal-polynomial theorem

Every reciprocal polynomial of even degree $2m$ can be written as a degree-$m$ polynomial in $x+x^{-1}$, multiplied by $x^m$. We test representative palindromic polynomials for $1\le m\le9$.

$$
F(x)=x^{2m}F(x^{-1})\ \Longrightarrow\ F(x)=x^mR(x+x^{-1}),\qquad \deg R=m.
$$

```python
print(v.check_general_reciprocal_reduction())
```

## 13. Separated blocks in arbitrary geometric products

For $m,q\ge2$ and $b\ge m$, different pairs $(r,s)$ produce different exponents $r+bs$. The test compares the support, coefficients and degree across a grid of parameters.

$$
\sum_{s=0}^{q-1}\sum_{r=0}^{m-1}x^{r+bs}=S_m(x)S_q(x^b),\quad \deg=m-1+b(q-1).
$$

```python
print(v.check_separated_exponent_blocks())
```

## 14. Products of equal-length sums

The overlap of the zero sets is governed by $g=\gcd(m,b)$. Roots shared by both factors are double, and the others are simple. We compare the theorem to exact square-free decompositions.

$$
N_{\rm double}=m-g,\quad N_{\rm distinct}=b(m-1)+g-1,\quad\deg P_{m,b}=(m-1)(b+1).
$$

```python
print(v.check_equal_length_counts())
```

## 15. Products of unequal-length sums

The intersection is the difference of the sets of roots of unity of orders dividing $\gcd(m,nb)$ and $\gcd(m,b)$. Exact symbolic counts are checked for $216$ parameter triples.

$$
\delta_{m,n,b}=\gcd(m,nb)-\gcd(m,b),
$$

$$
\nu_{m,n,b}=(m-1)+b(n-1)-\delta_{m,n,b}.
$$

```python
print(v.check_unequal_length_counts())
```

## 16. Real-root parity criterion

Every zero lies on the unit circle. Since $Q(1)=mn>0$, the only possible real zero is $-1$. Its multiplicity equals the number of factors vanishing there, determined by parity.

$$
\operatorname{mult}_{Q_{m,n,b}}(-1)=\mathbf1_{2\mid m}+\mathbf1_{2\mid n,\ 2\nmid b}.
$$

```python
print(v.check_real_root_parity())
```
