```python
from manim import *
import numpy as np
%load_ext manim

class HelixPerspective(ThreeDScene):
    def construct(self):
        self.camera.background_color = BLACK

        title = Text("Perspective Matters", font_size=36).to_edge(UP)
        self.add_fixed_in_frame_mobjects(title)

        axes = ThreeDAxes(
            x_range=[-4, 4, 1],
            y_range=[-2, 2, 1],
            z_range=[-2, 2, 1],
            x_length=7,
            y_length=3,
            z_length=3,
        )

        helix = ParametricFunction(
            lambda t: axes.c2p(
                t,
                np.cos(2 * t),
                np.sin(2 * t)
            ),
            t_range=[-3.5, 3.5],
            color=GREEN,
            stroke_width=5,
        )

        label = Text("Helix", font_size=24, color=YELLOW)
        label.to_edge(DOWN)
        self.add_fixed_in_frame_mobjects(label)

        self.set_camera_orientation(
            phi=65 * DEGREES,
            theta=-45 * DEGREES,
            zoom=0.8
        )

        self.play(Create(axes), Create(helix), run_time=3)
        self.wait(1)

        self.move_camera(
            phi=90 * DEGREES,
            theta=0 * DEGREES,
            zoom=0.9,
            run_time=4
        )

        label.become(
            Text("Cosine Wave", font_size=24, color=YELLOW).to_edge(DOWN)
        )

        self.wait(2)
```
